```python
from sympy import *
from sympy.plotting import plot

# rho: density
# k: Fermi momentum
x, s, k, rho = symbols('x s k rho', positive=True)
a, b = symbols('a b')

# spherical Bessel function of 1st order
j1 = sin(x)/x**2 - cos(x)/x
```

```python
series(9*(j1/x)**2, x, 0, 4)
```

```python
# exchange hole of the homogeneous electron gas
hole_hom = 9*rho*(j1.subs(x,k*s)/(k*s))**2
hole_hom
```

```python
series(hole_hom, s, 0, 4)
```

```python
# Gaussian exchange hole
hole = rho*(1+a*s**2+b*s**4)*exp(-(k**2/5) * s**2)
hole
```

```python
integrand = 4*pi*s**2 * hole
integral = integrate(integrand, (s,0,oo))
integral = simplify(integral)
integral
```

```python
# Integral of exchange hole should be 1
b_sol = solve(integral-1,b)[0]
b_sol
```

```python
series(hole, s,0,4)
```

```python
# exchange energy density
x_potential = integrate(4*pi * s * hole, (s,0,oo))
x_potential = simplify(x_potential)
x_potential
```

```python
# exchange energy density
xed = -rho/2*x_potential
xed
```

```python
# Express density in terms of Fermi momentum
collect(b_sol.subs(rho, k**3/(6*pi**2)), k)
```

```python
# Express Fermi momentum in terms of density
b_sol_rho = collect(b_sol.subs(k, (6*pi**2 * rho)**Rational(1,3)), rho)
b_sol_rho
```

```python
b_sol_rho.args[1].subs(a, 1).subs(rho, 1)
```
